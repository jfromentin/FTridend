```python
def quasi_shuffle(p,q):
    s=max(p,q)
    sum=0
    S=p+q
    for n in range(s,S+1):
        sum=sum+binom(s,n)*binom(S-n,s)
    return sum
```

```python
from sympy import poly, ZZ
from sympy.abc import x, y

# série des quasi-shuffles à partir un ordre plus grand strictement que 1
def quasi_shuffle_series(order):
    series=poly(1*x+1*y, domain=ZZ)
    for deg in range(2, order+1):
        for j in range(deg+1):
            series=series+quasi_shuffle(j,deg-j)*(x**j)*(y**(deg-j))
    return series
```
